```python

from pathlib import Path
from collections import Counter
import tensorflow as tf

dataset_path = Path("/content/civic_dataset")

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

# Use the exact category names found in your dataset.
categories = sorted([
    p.name for p in dataset_path.iterdir()
    if p.is_dir()
])

image_paths = {"train": [], "valid": [], "test": []}
image_labels = {"train": [], "valid": [], "test": []}

for label, category in enumerate(categories):
    category_root = dataset_path / category / category

    for split in ("train", "valid", "test"):
        image_folder = category_root / split / "images"

        if not image_folder.exists():
            print("WARNING: Missing folder:", image_folder)
            continue

        files = sorted(
            p for p in image_folder.iterdir()
            if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        )

        image_paths[split].extend(str(p) for p in files)
        image_labels[split].extend([label] * len(files))

print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))
print("\nCategories and label IDs:")
for i, category in enumerate(categories):
    print(f"{i}: {category}")

print("\nDataset summary:")
for split in ("train", "valid", "test"):
    counts = Counter(image_labels[split])
    print(f"{split}: {len(image_paths[split]):,} images")
    for i, category in enumerate(categories):
        print(f"  {category}: {counts[i]:,}")
```

```python

import json
import numpy as np
import tensorflow as tf
from pathlib import Path
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# Save results in Google Drive
model_dir = Path("/content/gdrive/MyDrive/CivicComplaintML/models")
model_dir.mkdir(parents=True, exist_ok=True)

IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 10
NUM_CLASSES = len(categories)
AUTOTUNE = tf.data.AUTOTUNE

# Load and preprocess each image
def load_image(path, label):
    image_bytes = tf.io.read_file(path)
    image = tf.io.decode_image(
        image_bytes, channels=3, expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = preprocess_input(image)
    return image, tf.cast(label, tf.int32)

def make_dataset(split, training=False):
    ds = tf.data.Dataset.from_tensor_slices(
        (image_paths[split], image_labels[split])
    )
    if training:
        ds = ds.shuffle(
            min(len(image_paths[split]), 10000),
            reshuffle_each_iteration=True
        )
    return (
        ds.map(load_image, num_parallel_calls=AUTOTUNE)
          .batch(BATCH_SIZE)
          .prefetch(AUTOTUNE)
    )

train_ds = make_dataset("train", training=True)
valid_ds = make_dataset("valid")
test_ds = make_dataset("test")

# Give minority categories additional weight during training
train_counts = np.bincount(
    image_labels["train"], minlength=NUM_CLASSES
)
total_train = len(image_labels["train"])

class_weights = {
    i: min(
        total_train / (NUM_CLASSES * max(int(train_counts[i]), 1)),
        5.0
    )
    for i in range(NUM_CLASSES)
}

# Create the model
base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)
base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.25)(x)
outputs = layers.Dense(
    NUM_CLASSES, activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Save the best model during training
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(model_dir / "best_civic_classifier.keras"),
        monitor="val_loss",
        save_best_only=True
    )
]

# Train the model
history = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks
)

# Evaluate on the separate test dataset
test_loss, test_accuracy = model.evaluate(test_ds)
print(f"Test accuracy: {test_accuracy:.2%}")
print(f"Test loss: {test_loss:.4f}")

# Save the final trained model and category names
model.save(model_dir / "civic_issue_classifier.keras")

with open(model_dir / "class_names.json", "w") as f:
    json.dump(categories, f, indent=2)

print("Training and saving complete!")
print("Saved files:")
for file in model_dir.iterdir():
    print(file.name)
```

```python

from collections import Counter
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

IMG_SIZE = 224
BATCH_SIZE = 32
VALID_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
splits = ["train", "valid", "test"]

# Find folders like .../train/images, .../valid/images, .../test/images
image_dirs = [
    p for p in DATASET_DIR.rglob("images")
    if p.is_dir() and p.parent.name in splits
]

if not image_dirs:
    raise RuntimeError(
        "No train/valid/test images folders found. "
        "Check the extracted dataset directory."
    )

# Find each category name among the folder's ancestors.
# The nearest matching category folder is used.
image_paths = {s: [] for s in splits}
image_labels = {s: [] for s in splits}

for img_dir in image_dirs:
    split = img_dir.parent.name
    category = next(
        (parent.name for parent in img_dir.parents
         if parent.name in class_names),
        None
    )
    if category is None:
        continue

    label = class_names.index(category)
    for file_path in img_dir.rglob("*"):
        if file_path.is_file() and file_path.suffix.lower() in VALID_EXTS:
            image_paths[split].append(str(file_path))
            image_labels[split].append(label)

# Remove duplicate paths if folders were discovered more than once.
for split in splits:
    pairs = sorted(set(zip(image_paths[split], image_labels[split])))
    image_paths[split] = [p for p, y in pairs]
    image_labels[split] = [y for p, y in pairs]

    counts = Counter(image_labels[split])
    print(f"\n{split}: {len(image_paths[split])} images")
    for i, name in enumerate(class_names):
        print(f"  {name}: {counts.get(i, 0)}")

    if not image_paths[split]:
        raise RuntimeError(f"No images found for split: {split}")

missing = [
    name for name in class_names
    if not any(
        class_names[y] == name
        for y in image_labels["train"]
    )
]
if missing:
    raise RuntimeError(f"Missing training images for: {missing}")

AUTOTUNE = tf.data.AUTOTUNE

def load_image(path, label, training=False):
    image_bytes = tf.io.read_file(path)
    image = tf.io.decode_image(
        image_bytes, channels=3, expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])

    # Augment only training images.
    if training:
        image = tf.image.random_flip_left_right(image)
        image = tf.image.random_brightness(image, max_delta=15.0)
        image = tf.image.random_contrast(image, lower=0.85, upper=1.15)

    image = preprocess_input(image)
    return image, label

def make_dataset(split, training=False):
    ds = tf.data.Dataset.from_tensor_slices(
        (image_paths[split], image_labels[split])
    )
    if training:
        ds = ds.shuffle(
            min(len(image_paths[split]), 5000),
            reshuffle_each_iteration=True
        )
    ds = ds.map(
        lambda p, y: load_image(p, y, training),
        num_parallel_calls=AUTOTUNE
    )
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)

train_ds = make_dataset("train", training=True)
valid_ds = make_dataset("valid")
test_ds = make_dataset("test")

print("\nAll datasets are ready.")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay
)

def evaluate_and_save(model_to_evaluate, dataset, prefix):
    y_true = np.concatenate([
        labels.numpy() for _, labels in dataset
    ])
    probabilities = model_to_evaluate.predict(dataset, verbose=1)
    y_pred = np.argmax(probabilities, axis=1)

    report_dict = classification_report(
        y_true, y_pred,
        labels=list(range(len(class_names))),
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )
    report_df = pd.DataFrame(report_dict).transpose()
    report_csv = RESULTS_DIR / f"{prefix}_classification_report.csv"
    report_df.to_csv(report_csv)

    cm = confusion_matrix(
        y_true, y_pred, labels=list(range(len(class_names)))
    )
    cm_df = pd.DataFrame(cm, index=class_names, columns=class_names)
    cm_df.to_csv(RESULTS_DIR / f"{prefix}_confusion_matrix.csv")

    fig, ax = plt.subplots(figsize=(13, 11))
    ConfusionMatrixDisplay(
        confusion_matrix=cm, display_labels=class_names
    ).plot(ax=ax, xticks_rotation=90, colorbar=False, values_format="d")
    fig.tight_layout()
    fig.savefig(
        RESULTS_DIR / f"{prefix}_confusion_matrix.png",
        dpi=160, bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)

    accuracy = float(np.mean(y_true == y_pred))
    print(f"\n{prefix} accuracy: {accuracy:.4f}")
    print(report_df.loc[class_names, ["precision", "recall", "f1-score", "support"]])
    print("\nSaved report and confusion matrix in:", RESULTS_DIR)

    return y_true, y_pred, probabilities, report_df

baseline_true, baseline_pred, baseline_probs, baseline_report = \
    evaluate_and_save(model, test_ds, "baseline")
```

```python

import numpy as np
import tensorflow as tf
from sklearn.utils.class_weight import compute_class_weight
from pathlib import Path

IMPROVED_MODEL_PATH = MODEL_DIR / "civic_classifier_finetuned.keras"

# Calculate class weights using training labels only.
labels = np.asarray(image_labels["train"], dtype=np.int32)
classes = np.arange(len(class_names))

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=labels
)

# Cap weights so very rare classes do not dominate training.
class_weights = {
    int(i): float(min(w, 4.0))
    for i, w in enumerate(weights)
}

print("Class weights:")
for i, name in enumerate(class_names):
    print(f"{name}: {class_weights[i]:.2f}")

# Locate the nested MobileNetV2 model.
base_candidates = [
    layer for layer in model.layers
    if isinstance(layer, tf.keras.Model)
    and "mobilenet" in layer.name.lower()
]

if not base_candidates:
    raise RuntimeError(
        "Could not find the MobileNetV2 base model. "
        "Inspect model.summary() before proceeding."
    )

base_model = base_candidates[0]
base_model.trainable = True

# Fine-tune only the last 30 base-model layers.
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep Batch Normalization layers frozen for stability.
for layer in base_model.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(IMPROVED_MODEL_PATH),
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=1,
        min_lr=1e-7,
        verbose=1
    )
]

history = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=8,
    class_weight=class_weights,
    callbacks=callbacks
)

# Save the final restored weights separately too.
model.save(str(MODEL_DIR / "civic_classifier_finetuned_final.keras"))

print("\nFine-tuning complete.")
print("Best checkpoint:", IMPROVED_MODEL_PATH)
print("Original model still exists:", (MODEL_DIR / "best_civic_classifier.keras").exists())
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import json
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

BEST_MODEL_PATH = MODEL_DIR / "best_civic_classifier.keras"
IMPROVED_MODEL_PATH = MODEL_DIR / "civic_classifier_finetuned.keras"

# Load the best checkpoint selected by validation loss.
baseline_model = tf.keras.models.load_model(BEST_MODEL_PATH)
improved_model = tf.keras.models.load_model(IMPROVED_MODEL_PATH)

def evaluate_and_save(model_to_evaluate, dataset, prefix):
    # test_ds must not be shuffled.
    y_true = np.concatenate([
        y.numpy() for _, y in dataset
    ])

    probabilities = model_to_evaluate.predict(dataset, verbose=1)
    y_pred = np.argmax(probabilities, axis=1)

    report = classification_report(
        y_true,
        y_pred,
        labels=list(range(len(class_names))),
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    report_df = pd.DataFrame(report).transpose()
    report_df.to_csv(
        RESULTS_DIR / f"{prefix}_classification_report.csv"
    )

    cm = confusion_matrix(
        y_true, y_pred,
        labels=list(range(len(class_names)))
    )

    pd.DataFrame(
        cm, index=class_names, columns=class_names
    ).to_csv(RESULTS_DIR / f"{prefix}_confusion_matrix.csv")

    fig, ax = plt.subplots(figsize=(13, 11))
    ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=class_names
    ).plot(
        ax=ax,
        xticks_rotation=90,
        values_format="d",
        colorbar=False
    )
    fig.tight_layout()
    fig.savefig(
        RESULTS_DIR / f"{prefix}_confusion_matrix.png",
        dpi=160,
        bbox_inches="tight"
    )
    plt.show()
    plt.close(fig)

    accuracy = float(np.mean(y_true == y_pred))
    print(f"\n{prefix} test accuracy: {accuracy:.4%}")
    print("\nPer-class metrics:")
    print(report_df.loc[
        class_names, ["precision", "recall", "f1-score", "support"]
    ].round(4))

    return y_true, y_pred, probabilities, report_df, accuracy

# Evaluate the original model.
base_true, base_pred, base_probs, base_report, base_accuracy = \
    evaluate_and_save(baseline_model, test_ds, "baseline")

# Evaluate the fine-tuned model.
fine_true, fine_pred, fine_probs, fine_report, fine_accuracy = \
    evaluate_and_save(improved_model, test_ds, "finetuned")

# Save a compact comparison.
comparison = pd.DataFrame({
    "model": ["baseline", "fine_tuned"],
    "test_accuracy": [base_accuracy, fine_accuracy],
    "pothole_precision": [
        base_report.loc["Potholes and RoadCracks", "precision"],
        fine_report.loc["Potholes and RoadCracks", "precision"]
    ],
    "pothole_recall": [
        base_report.loc["Potholes and RoadCracks", "recall"],
        fine_report.loc["Potholes and RoadCracks", "recall"]
    ],
    "pothole_f1": [
        base_report.loc["Potholes and RoadCracks", "f1-score"],
        fine_report.loc["Potholes and RoadCracks", "f1-score"]
    ]
})

comparison.to_csv(
    RESULTS_DIR / "model_comparison.csv", index=False
)

print("\nMODEL COMPARISON")
print(comparison.round(4).to_string(index=False))
print("\nAll reports saved to:", RESULTS_DIR)
```

```python
!pip install -q pillow

import json
import numpy as np
import tensorflow as tf
from PIL import Image
from google.colab import files
from IPython.display import display

MODEL_PATH = "/content/gdrive/MyDrive/CivicComplaintML/models/best_civic_classifier.keras"
CLASSES_PATH = "/content/gdrive/MyDrive/CivicComplaintML/models/class_names.json"

model = tf.keras.models.load_model(MODEL_PATH)

with open(CLASSES_PATH, "r") as f:
    class_names = json.load(f)

print("Model loaded successfully!")
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
print("Number of classes:", len(class_names))
print("Classes:", class_names)
```

```python
import pandas as pd
from pathlib import Path

PROJECT = Path("/content/gdrive/MyDrive/CivicComplaintML")
RESULTS = PROJECT / "results"

for name in ["baseline", "finetuned"]:
    print("\n" + "=" * 60)
    print(f"{name.upper()} MODEL")
    print("=" * 60)

    report_path = RESULTS / f"{name}_classification_report.csv"
    matrix_path = RESULTS / f"{name}_confusion_matrix.csv"

    print("\n--- Classification Report ---")
    report = pd.read_csv(report_path, index_col=0)
    print(report.to_string())

    print("\n--- Confusion Matrix ---")
    matrix = pd.read_csv(matrix_path, index_col=0)
    print(matrix.to_string())

    # Find the graffiti row in the classification report
    graffiti_rows = [
        idx for idx in report.index
        if "graff" in str(idx).lower()
    ]

    for idx in graffiti_rows:
        row = report.loc[idx]
        print(f"\nGraffiti metrics ({idx}):")
        for metric in ["precision", "recall", "f1-score", "support"]:
            if metric in row.index:
                print(f"  {metric}: {row[metric]}")

    # Show actual classes most often predicted as graffiti
    graffiti_cols = [
        col for col in matrix.columns
        if "graff" in str(col).lower()
    ]

    if graffiti_cols:
        col = graffiti_cols[0]
        print(f"\nActual classes predicted as {col}:")
        print(matrix[col].sort_values(ascending=False).to_string())
```

```python
import numpy as np
from PIL import Image
from google.colab import files
from IPython.display import display

GRAFFITI_THRESHOLD = 0.70  # Initial value to evaluate, not yet validated

uploaded = files.upload()

for filename in uploaded:
    image = Image.open(filename).convert("RGB")
    display(image)

    image = image.resize((224, 224))
    image_array = np.asarray(image, dtype=np.float32)
    image_array = np.expand_dims(image_array, axis=0)

    predictions = model.predict(image_array, verbose=0)[0]

    graffiti_index = next(
        i for i, name in enumerate(class_names)
        if "graff" in name.lower()
    )

    top_index = int(np.argmax(predictions))
    graffiti_score = float(predictions[graffiti_index])

    if top_index == graffiti_index and graffiti_score < GRAFFITI_THRESHOLD:
        # Do not confidently label this as graffiti.
        alternatives = predictions.copy()
        alternatives[graffiti_index] = -1
        alternative_index = int(np.argmax(alternatives))

        print("Graffiti prediction rejected by threshold.")
        print("Possible alternative:", class_names[alternative_index])
        print(f"Graffiti score: {graffiti_score:.2%}")
        print("Note: the alternative may also be incorrect.")

    else:
        print("Predicted issue:", class_names[top_index])
        print(f"Model score: {predictions[top_index]:.2%}")
```

```python
from pathlib import Path

ROOT = Path("/content/gdrive/MyDrive")

image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
skip_dirs = {
    ".ipynb_checkpoints", "node_modules", ".git",
    "urbanpulse_ai"
}

print("Searching Google Drive for image datasets...\n")

found = 0

for folder in ROOT.iterdir():
    if not folder.is_dir() or folder.name.startswith("."):
        continue

    # Look for directories with class subfolders containing images.
    for directory in folder.rglob("*"):
        if not directory.is_dir():
            continue

        if any(part in skip_dirs for part in directory.parts):
            continue

        try:
            subdirs = [p for p in directory.iterdir() if p.is_dir()]
            counts = []

            for subdir in subdirs:
                n = sum(
                    1 for p in subdir.rglob("*")
                    if p.is_file() and p.suffix.lower() in image_exts
                )
                if n:
                    counts.append((subdir.name, n))

            if len(counts) >= 3:
                print(f"\nDataset candidate: {directory}")
                for name, count in counts:
                    print(f"  {name}: {count} images")
                found += 1

        except (PermissionError, OSError):
            continue

        if found >= 10:
            break

    if found >= 10:
        break

if found == 0:
    print("No obvious class-folder dataset found.")
```

```python

from pathlib import Path
import shutil

SOURCE = Path('/content/urban_issues_dataset')
DEST = Path('/content/urbanpulse_8class')

IMAGE_EXTS = {'.jpg', '.jpeg', '.png', '.webp', '.bmp'}
SPLIT_MAP = {'train': 'train', 'valid': 'val', 'test': 'test'}

if DEST.exists():
    raise RuntimeError(
        f"{DEST} already exists. Stop here so we don't overwrite an experiment."
    )

DEST.mkdir(parents=True)

class_folders = sorted(
    p for p in SOURCE.iterdir()
    if p.is_dir() and p.name.lower() not in {'graffitti', 'graffiti'}
)

for class_folder in class_folders:
    for source_split, target_split in SPLIT_MAP.items():
        image_dirs = [
            p for p in class_folder.rglob('images')
            if p.parent.name == source_split
        ]

        for image_dir in image_dirs:
            for img in image_dir.iterdir():
                if not img.is_file() or img.suffix.lower() not in IMAGE_EXTS:
                    continue

                target_dir = DEST / target_split / class_folder.name
                target_dir.mkdir(parents=True, exist_ok=True)

                # Avoid filename collisions between source datasets.
                target_name = f'{class_folder.name}__{img.name}'
                shutil.copy2(img, target_dir / target_name)

print("Prepared dataset:", DEST)

for split in ['train', 'val', 'test']:
    print(f'\n{split}:')
    split_dir = DEST / split
    for cls in sorted(p for p in split_dir.iterdir() if p.is_dir()):
        print(f'  {cls.name}: {sum(1 for f in cls.iterdir() if f.is_file())}')
```

```python

import tensorflow as tf
from pathlib import Path

DATA = Path('/content/urbanpulse_8class')
MODEL_OUT = Path('/content/urbanpulse_8class_mobilenetv2.keras')

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA / 'train',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    seed=SEED,
    label_mode='int'
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA / 'val',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,
    label_mode='int'
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    DATA / 'test',
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False,
    label_mode='int'
)

class_names = train_ds.class_names
print("Classes:", class_names)
assert len(class_names) == 8, f"Expected 8 classes, got {len(class_names)}"

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

base = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base.trainable = False

inputs = tf.keras.Input(shape=(224, 224, 3))
x = tf.keras.applications.mobilenet_v2.preprocess_input(inputs)
x = base(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.25)(x)
outputs = tf.keras.layers.Dense(len(class_names), activation='softmax')(x)

model = tf.keras.Model(inputs, outputs)
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=3, restore_best_weights=True
    ),
    tf.keras.callbacks.ModelCheckpoint(
        '/content/urbanpulse_8class_best.keras',
        monitor='val_loss',
        save_best_only=True
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=12,
    callbacks=callbacks
)

model = tf.keras.models.load_model('/content/urbanpulse_8class_best.keras')
test_loss, test_accuracy = model.evaluate(test_ds, verbose=1)

print(f'Test accuracy: {test_accuracy:.4f}')
model.save(MODEL_OUT)

import json
Path('/content/urbanpulse_8class_class_names.json').write_text(
    json.dumps(class_names, indent=2)
)

print("Saved model:", MODEL_OUT)
print("Saved class names: /content/urbanpulse_8class_class_names.json")
```

```python
from google.colab import drive, files
drive.mount('/content/gdrive')

import tensorflow as tf
import numpy as np
import json
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

MODEL_PATH = "/content/gdrive/MyDrive/CivicComplaintML/models/urbanpulse_8class_pothole_stage1.keras"
CLASS_PATH = "/content/gdrive/MyDrive/CivicComplaintML/models/urbanpulse_8class_class_names.json"

model = tf.keras.models.load_model(MODEL_PATH)

with open(CLASS_PATH, "r") as f:
    class_names = json.load(f)

print("Refined model loaded.")
print("Available classes:", class_names)
```

```python
CONFIDENCE_THRESHOLD = 0.70
MARGIN_THRESHOLD = 0.15

UNKNOWN_LABEL = "Does not belong to any of the 8 classes"

print("Unknown-image detection enabled.")
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

results = []

for filename in uploaded.keys():
    try:
        img = Image.open(filename).convert("RGB")
        resized_img = img.resize((224, 224))

        image_array = np.array(resized_img, dtype=np.float32)
        image_array = np.expand_dims(image_array, axis=0)

        # Match MobileNetV2 preprocessing used during training
        image_array = preprocess_input(image_array)

        scores = model.predict(image_array, verbose=0)[0]

        sorted_indices = np.argsort(scores)[::-1]

        top_index = int(sorted_indices[0])
        second_index = int(sorted_indices[1])

        top_confidence = float(scores[top_index])
        second_confidence = float(scores[second_index])
        confidence_margin = top_confidence - second_confidence

        if (
            top_confidence >= CONFIDENCE_THRESHOLD
            and confidence_margin >= MARGIN_THRESHOLD
        ):
            prediction = class_names[top_index]
            status = "KNOWN_ISSUE"
        else:
            prediction = UNKNOWN_LABEL
            status = "ADMIN_REVIEW"

        results.append({
            "Photo": filename,
            "Prediction": prediction,
            "Status": status,
            "Top confidence (%)": round(top_confidence * 100, 2),
            "Second class": class_names[second_index],
            "Second confidence (%)": round(second_confidence * 100, 2),
            "Confidence margin (%)": round(confidence_margin * 100, 2)
        })

        plt.figure(figsize=(6, 4))
        plt.imshow(img)
        plt.axis("off")
        plt.title(
            f"{prediction}\n"
            f"Confidence: {top_confidence * 100:.2f}%"
        )
        plt.show()

        print("Photo:", filename)
        print("Prediction:", prediction)
        print("Status:", status)
        print("Top confidence:", round(top_confidence * 100, 2), "%")
        print("-" * 50)

    except Exception as e:
        print(f"Error processing {filename}: {e}")
```

```python
import os
import shutil

source_root = "/content/urbanpulse_civic_source"
positive_dest = "/content/urbanpulse_binary/CIVIC_ISSUE"

classes = [
    "Damaged concrete structures",
    "DamagedElectricalPoles",
    "DamagedRoadSigns",
    "DeadAnimalsPollution",
    "FallenTrees",
    "Garbage",
    "IllegalParking",
    "Potholes and RoadCracks"
]

extensions = (".jpg", ".jpeg", ".png", ".bmp")

# Clear the previous empty destination
if os.path.exists(positive_dest):
    shutil.rmtree(positive_dest)

os.makedirs(positive_dest, exist_ok=True)

copied = 0

for class_name in classes:

    image_folder = os.path.join(
        source_root,
        class_name,
        class_name,
        "train",
        "images"
    )

    if not os.path.isdir(image_folder):
        print("Missing:", image_folder)
        continue

    class_count = 0

    for filename in os.listdir(image_folder):

        if filename.lower().endswith(extensions):

            source = os.path.join(image_folder, filename)

            destination = os.path.join(
                positive_dest,
                f"{class_name.replace(' ', '_')}_{copied}{os.path.splitext(filename)[1]}"
            )

            shutil.copy2(source, destination)

            copied += 1
            class_count += 1

    print(f"{class_name}: copied {class_count}")

print("\nTotal positive images copied:", copied)
```

```python
import os
import shutil
import random

random.seed(42)

binary_root = "/content/urbanpulse_binary"

positive_train_source = os.path.join(binary_root, "CIVIC_ISSUE")
negative_source = os.path.join(binary_root, "NOT_CIVIC_ISSUE")

classes = [
    "Damaged concrete structures",
    "DamagedElectricalPoles",
    "DamagedRoadSigns",
    "DeadAnimalsPollution",
    "FallenTrees",
    "Garbage",
    "IllegalParking",
    "Potholes and RoadCracks"
]

extensions = (".jpg", ".jpeg", ".png", ".bmp")

# Collect original validation images
positive_val_candidates = []

for class_name in classes:
    folder = os.path.join(
        "/content/urbanpulse_civic_source",
        class_name,
        class_name,
        "valid",
        "images"
    )

    if os.path.isdir(folder):
        for filename in os.listdir(folder):
            if filename.lower().endswith(extensions):
                positive_val_candidates.append(
                    os.path.join(folder, filename)
                )

# Collect existing images
positive_train_candidates = [
    os.path.join(positive_train_source, f)
    for f in os.listdir(positive_train_source)
    if f.lower().endswith(extensions)
]

negative_candidates = [
    os.path.join(negative_source, f)
    for f in os.listdir(negative_source)
    if f.lower().endswith(extensions)
]

print("Civic training candidates:", len(positive_train_candidates))
print("Civic validation candidates:", len(positive_val_candidates))
print("Negative candidates:", len(negative_candidates))

# Choose balanced counts
train_count = 8000
val_count = min(2000, len(positive_val_candidates),
                len(negative_candidates) - train_count)

assert len(positive_train_candidates) >= train_count
assert val_count > 0

# Select images without duplication
positive_train_selected = random.sample(
    positive_train_candidates, train_count
)
positive_val_selected = random.sample(
    positive_val_candidates, val_count
)

negative_selected = random.sample(
    negative_candidates, train_count + val_count
)

negative_train_selected = negative_selected[:train_count]
negative_val_selected = negative_selected[train_count:]

# Create clean output structure
prepared_root = "/content/urbanpulse_binary_prepared"

if os.path.exists(prepared_root):
    shutil.rmtree(prepared_root)

for split in ["train", "val"]:
    for label in ["CIVIC_ISSUE", "NOT_CIVIC_ISSUE"]:
        os.makedirs(
            os.path.join(prepared_root, split, label),
            exist_ok=True
        )

def copy_images(paths, destination, prefix):
    for i, source in enumerate(paths):
        ext = os.path.splitext(source)[1].lower()
        shutil.copy2(
            source,
            os.path.join(destination, f"{prefix}_{i}{ext}")
        )

copy_images(
    positive_train_selected,
    os.path.join(prepared_root, "train", "CIVIC_ISSUE"),
    "civic_train"
)

copy_images(
    negative_train_selected,
    os.path.join(prepared_root, "train", "NOT_CIVIC_ISSUE"),
    "negative_train"
)

copy_images(
    positive_val_selected,
    os.path.join(prepared_root, "val", "CIVIC_ISSUE"),
    "civic_val"
)

copy_images(
    negative_val_selected,
    os.path.join(prepared_root, "val", "NOT_CIVIC_ISSUE"),
    "negative_val"
)

print("\nPrepared dataset:")
for split in ["train", "val"]:
    for label in ["CIVIC_ISSUE", "NOT_CIVIC_ISSUE"]:
        folder = os.path.join(prepared_root, split, label)
        print(split, label, len(os.listdir(folder)))
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/urbanpulse_binary_prepared/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/urbanpulse_binary_prepared/val",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Class order:", train_ds.class_names)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# Load pretrained MobileNetV2 feature extractor
base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze pretrained layers for initial training
base_model.trainable = False

# Build binary classifier
inputs = layers.Input(shape=(224, 224, 3))

x = layers.Lambda(preprocess_input)(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    1,
    activation="sigmoid",
    name="civic_relevance"
)(x)

binary_model = models.Model(inputs, outputs)

binary_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

binary_model.summary()
```

```python
import os
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

model_folder = (
    "/content/gdrive/MyDrive/CivicComplaintML/models"
)

os.makedirs(model_folder, exist_ok=True)

binary_model_path = os.path.join(
    model_folder,
    "urbanpulse_binary_relevance.keras"
)

callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        filepath=binary_model_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]

print("Binary model will be saved at:")
print(binary_model_path)
```

```python
history = binary_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks
)

print("\nBinary classifier training completed!")
print("Saved model:", binary_model_path)
```

```python
import os
import glob
import pandas as pd
from PIL import Image, ImageDraw
from tqdm import tqdm
import math

BASE = "/content/street_surface_vis"
IMAGE_DIR = "/content/street_surface_vis/images"

# Find extracted image files
image_extensions = (".jpg", ".jpeg", ".png", ".webp")

all_image_paths = []

for root, dirs, files in os.walk(IMAGE_DIR):
    for filename in files:
        if filename.lower().endswith(image_extensions):
            all_image_paths.append(os.path.join(root, filename))

# Also inspect the actual extraction directory if needed
if not all_image_paths:
    for root, dirs, files in os.walk("/content/street_surface_vis"):
        for filename in files:
            if filename.lower().endswith(image_extensions):
                all_image_paths.append(os.path.join(root, filename))

print("Extracted image count:", len(all_image_paths))

# Read labels
labels = pd.read_csv(os.path.join(BASE, "labels.csv"))

# Map image IDs to actual files
image_map = {
    os.path.splitext(os.path.basename(path))[0]: path
    for path in all_image_paths
}

labels["image_path"] = labels["mapillary_image_id"].astype(str).map(image_map)

print("Images matched to CSV:", labels["image_path"].notna().sum())

# Candidate negatives: good or excellent road surfaces
candidates = labels[
    labels["surface_quality"].astype(str).str.lower().isin(["good", "excellent"])
].copy()

candidates = candidates[candidates["image_path"].notna()]

print("\nCandidate quality distribution:")
print(candidates["surface_quality"].value_counts())

print("\nCandidate images:", len(candidates))

# Create contact sheets
REVIEW_DIR = "/content/urbanpulse_street_review"
os.makedirs(REVIEW_DIR, exist_ok=True)

candidate_paths = candidates["image_path"].tolist()

random.seed(42)
random.shuffle(candidate_paths)

SHEET_SIZE = 40
THUMB_W, THUMB_H = 180, 125

for start in range(0, min(len(candidate_paths), 400), SHEET_SIZE):
    batch = candidate_paths[start:start + SHEET_SIZE]

    cols = 5
    rows = math.ceil(len(batch) / cols)

    sheet = Image.new(
        "RGB",
        (cols * THUMB_W, rows * (THUMB_H + 25)),
        "white"
    )

    draw = ImageDraw.Draw(sheet)

    for i, path in enumerate(batch):
        try:
            img = Image.open(path).convert("RGB")
            img.thumbnail((THUMB_W, THUMB_H))

            x = (i % cols) * THUMB_W
            y = (i // cols) * (THUMB_H + 25)

            sheet.paste(img, (x, y))
            draw.text(
                (x + 3, y + THUMB_H),
                str(start + i),
                fill="black"
            )
        except Exception:
            continue

    sheet.save(
        os.path.join(REVIEW_DIR, f"sheet_{start // SHEET_SIZE:02d}.jpg")
    )

print("\nContact sheets created:", REVIEW_DIR)
print("Number of sheets:", math.ceil(min(len(candidate_paths), 400) / SHEET_SIZE))
```

```python
import os
import tensorflow as tf
import numpy as np

# Save locally first; copy to Drive after training
LOCAL_CHECKPOINT = "/content/urbanpulse_binary_urban_v2_best.keras"

EPOCHS = 8
PATIENCE = 2
best_val_loss = float("inf")
wait = 0

def calculate_metrics(losses, correct, total):
    return (
        float(np.mean(losses)),
        float(correct / total) if total else 0.0
    )

for epoch in range(EPOCHS):

    print(f"\n{'='*45}")
    print(f"Epoch {epoch + 1}/{EPOCHS}")
    print(f"{'='*45}")

    # -------------------------
    # TRAINING
    # -------------------------
    train_losses = []
    train_correct = 0
    train_total = 0

    for images, labels in train_ds:

        labels = tf.reshape(tf.cast(labels, tf.float32), [-1, 1])

        # Update model weights
        binary_model_v2.train_on_batch(images, labels)

        # Calculate metrics without model.evaluate()
        predictions = binary_model_v2(images, training=False)

        batch_loss = tf.reduce_mean(
            tf.keras.losses.binary_crossentropy(labels, predictions)
        )

        predicted_labels = tf.cast(predictions >= 0.5, tf.float32)

        train_losses.append(float(batch_loss.numpy()))
        train_correct += int(
            tf.reduce_sum(
                tf.cast(tf.equal(predicted_labels, labels), tf.int32)
            ).numpy()
        )
        train_total += int(tf.size(labels).numpy())

    train_loss, train_accuracy = calculate_metrics(
        train_losses, train_correct, train_total
    )

    # -------------------------
    # VALIDATION
    # -------------------------
    val_losses = []
    val_correct = 0
    val_total = 0

    for images, labels in val_ds:

        labels = tf.reshape(tf.cast(labels, tf.float32), [-1, 1])

        predictions = binary_model_v2(images, training=False)

        batch_loss = tf.reduce_mean(
            tf.keras.losses.binary_crossentropy(labels, predictions)
        )

        predicted_labels = tf.cast(predictions >= 0.5, tf.float32)

        val_losses.append(float(batch_loss.numpy()))
        val_correct += int(
            tf.reduce_sum(
                tf.cast(tf.equal(predicted_labels, labels), tf.int32)
            ).numpy()
        )
        val_total += int(tf.size(labels).numpy())

    val_loss, val_accuracy = calculate_metrics(
        val_losses, val_correct, val_total
    )

    print(f"Training loss:      {train_loss:.4f}")
    print(f"Training accuracy:  {train_accuracy:.4f}")
    print(f"Validation loss:    {val_loss:.4f}")
    print(f"Validation accuracy:{val_accuracy:.4f}")

    # -------------------------
    # SAVE BEST CHECKPOINT
    # -------------------------
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        wait = 0

        binary_model_v2.save(LOCAL_CHECKPOINT)

        print("Best checkpoint saved locally.")

    else:
        wait += 1
        print(f"No validation-loss improvement ({wait}/{PATIENCE}).")

    if wait >= PATIENCE:
        print("Early stopping.")
        break

print("\nTraining loop completed.")
print("Best validation loss:", best_val_loss)
print("Checkpoint exists:", os.path.exists(LOCAL_CHECKPOINT))
```

```python
import os
import tensorflow as tf
from PIL import Image

DATA_DIR = "/content/urbanpulse_binary_v2_data"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

class_names = ["CIVIC_ISSUE", "NOT_CIVIC_ISSUE"]

def get_valid_files(folder):
    paths = []

    for filename in os.listdir(folder):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):
            path = os.path.join(folder, filename)

            try:
                with Image.open(path) as img:
                    img.verify()
                paths.append(path)
            except Exception:
                print("Skipping invalid:", filename)

    return paths


def make_dataset(split, shuffle=False):
    paths, labels = [], []

    for label, class_name in enumerate(class_names):
        folder = os.path.join(DATA_DIR, split, class_name)

        if not os.path.isdir(folder):
            raise FileNotFoundError(f"Missing folder: {folder}")

        files = get_valid_files(folder)
        paths.extend(files)
        labels.extend([label] * len(files))

    if not paths:
        raise ValueError(f"No valid images found in {split}")

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))

    if shuffle:
        ds = ds.shuffle(len(paths), seed=42)

    def load_image(path, label):
        content = tf.io.read_file(path)

        is_png = tf.strings.regex_full_match(
            tf.strings.lower(path), ".*\\.png"
        )

        image = tf.cond(
            is_png,
            lambda: tf.io.decode_png(content, channels=3),
            lambda: tf.io.decode_jpeg(content, channels=3)
        )

        image = tf.image.resize(image, IMG_SIZE)
        image = tf.ensure_shape(image, [224, 224, 3])

        return tf.cast(image, tf.float32), tf.cast(label, tf.float32)

    ds = ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

    print(split, "images:", len(paths))
    return ds


train_ds = make_dataset("train", shuffle=True)
val_ds = make_dataset("val", shuffle=False)

print("Datasets restored successfully.")
```

```python
import os
import shutil
import random
import pandas as pd
from pathlib import Path
from PIL import Image

# --------------------------------------------------
# 1. Paths and settings
# --------------------------------------------------

SOURCE_DIR = Path("/content/urban_issues_dataset_recovered")
MANIFEST_PATH = Path(
    "/content/gdrive/MyDrive/CivicComplaintML/results/"
    "street_surface_vis_candidates.csv"
)

OUTPUT_DIR = Path("/content/urbanpulse_binary_v2_data")

CLASSES = [
    "Damaged concrete structures",
    "DamagedElectricalPoles",
    "DamagedRoadSigns",
    "DeadAnimalsPollution",
    "FallenTrees",
    "Garbage",
    "IllegalParking",
    "Potholes and RoadCracks"
]

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
random.seed(42)

# --------------------------------------------------
# 2. Find civic images in original train/valid folders
# --------------------------------------------------

civic_train = []
civic_valid = []

for class_name in CLASSES:
    class_dir = SOURCE_DIR / class_name

    for path in class_dir.rglob("*"):
        if not path.is_file() or path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        parts = [part.lower() for part in path.parts]

        if "train" in parts:
            civic_train.append(path)
        elif "valid" in parts or "val" in parts:
            civic_valid.append(path)

print("Civic training images found:", len(civic_train))
print("Civic validation images found:", len(civic_valid))

# --------------------------------------------------
# 3. Load clean-road candidates from saved manifest
# --------------------------------------------------

negative_df = pd.read_csv(MANIFEST_PATH)
negative_paths = [
    Path(p) for p in negative_df["image_path"].dropna().tolist()
]

print("Negative candidate images found:", len(negative_paths))

# --------------------------------------------------
# 4. Verify images and remove unreadable files
# --------------------------------------------------

def valid_images(paths):
    good = []

    for path in paths:
        try:
            with Image.open(path) as img:
                img.verify()
            good.append(path)
        except Exception:
            pass

    return good

civic_train = valid_images(civic_train)
civic_valid = valid_images(civic_valid)
negative_paths = valid_images(negative_paths)

print("\nReadable images:")
print("Civic train:", len(civic_train))
print("Civic validation:", len(civic_valid))
print("Clean-road candidates:", len(negative_paths))

# --------------------------------------------------
# 5. Check sufficient image counts
# --------------------------------------------------

assert len(civic_train) >= 4000, "Not enough civic training images."
assert len(civic_valid) >= 1000, "Not enough civic validation images."
assert len(negative_paths) >= 4900, "Not enough negative images."

# --------------------------------------------------
# 6. Select balanced samples
# --------------------------------------------------

random.shuffle(civic_train)
random.shuffle(civic_valid)
random.shuffle(negative_paths)

selected = {
    "train/CIVIC_ISSUE": civic_train[:4000],
    "val/CIVIC_ISSUE": civic_valid[:1000],
    "train/NOT_CIVIC_ISSUE": negative_paths[:4000],
    "val/NOT_CIVIC_ISSUE": negative_paths[4000:4900]
}

# --------------------------------------------------
# 7. Create output folders and copy images
# --------------------------------------------------

if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)

for split_class, paths in selected.items():
    destination = OUTPUT_DIR / split_class
    destination.mkdir(parents=True, exist_ok=True)

    for index, source_path in enumerate(paths):
        filename = f"image_{index:05d}{source_path.suffix.lower()}"
        shutil.copy2(source_path, destination / filename)

# --------------------------------------------------
# 8. Print final dataset counts
# --------------------------------------------------

print("\nDATASET PREPARATION COMPLETE")

for split in ["train", "val"]:
    for class_name in ["CIVIC_ISSUE", "NOT_CIVIC_ISSUE"]:
        folder = OUTPUT_DIR / split / class_name
        count = sum(1 for p in folder.iterdir() if p.is_file())
        print(f"{split}/{class_name}: {count}")

print("\nDataset location:", OUTPUT_DIR)
```

```python
import os
import shutil
import tensorflow as tf
import numpy as np

# --------------------------------------------------
# 1. Training configuration
# --------------------------------------------------

EPOCHS = 8
LEARNING_RATE = 1e-4
THRESHOLD = 0.5
PATIENCE = 3

LOCAL_BEST = "/content/urbanpulse_binary_v2_best.keras"
DRIVE_BEST = (
    "/content/gdrive/MyDrive/CivicComplaintML/models/"
    "urbanpulse_binary_v2_best.keras"
)

optimizer = tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE)
bce = tf.keras.losses.BinaryCrossentropy()

# --------------------------------------------------
# 2. Training step
# --------------------------------------------------

@tf.function
def train_step(images, labels):
    with tf.GradientTape() as tape:
        probabilities = binary_model_v2(images, training=True)
        loss = bce(labels, probabilities)

    gradients = tape.gradient(
        loss, binary_model_v2.trainable_variables
    )

    optimizer.apply_gradients(
        zip(gradients, binary_model_v2.trainable_variables)
    )

    return loss

# --------------------------------------------------
# 3. Validation step
# --------------------------------------------------

def validate_model():
    total_loss = 0.0
    total_samples = 0

    civic_correct = 0
    civic_total = 0
    noncivic_correct = 0
    noncivic_total = 0

    for images, labels in val_ds:
        probabilities = binary_model_v2(images, training=False)

        loss = bce(labels, probabilities)
        batch_size = int(tf.shape(labels)[0])

        total_loss += float(loss) * batch_size
        total_samples += batch_size

        # Output >= 0.5 means NOT_CIVIC_ISSUE
        predicted_noncivic = probabilities >= THRESHOLD
        actual_noncivic = labels >= 0.5

        # Civic class has label 0
        civic_mask = tf.logical_not(actual_noncivic)
        civic_correct += int(tf.reduce_sum(tf.cast(
            tf.logical_and(
                civic_mask,
                tf.logical_not(predicted_noncivic)
            ), tf.int32
        )))
        civic_total += int(tf.reduce_sum(tf.cast(civic_mask, tf.int32)))

        # Non-civic class has label 1
        noncivic_correct += int(tf.reduce_sum(tf.cast(
            tf.logical_and(actual_noncivic, predicted_noncivic),
            tf.int32
        )))
        noncivic_total += int(tf.reduce_sum(tf.cast(
            actual_noncivic, tf.int32
        )))

    return {
        "loss": total_loss / total_samples,
        "civic_recall": civic_correct / max(civic_total, 1),
        "noncivic_recall": noncivic_correct / max(noncivic_total, 1)
    }

# --------------------------------------------------
# 4. Train with early stopping
# --------------------------------------------------

best_val_loss = float("inf")
epochs_without_improvement = 0

for epoch in range(1, EPOCHS + 1):
    training_loss = 0.0
    training_samples = 0

    for images, labels in train_ds:
        loss = train_step(images, labels)
        batch_size = int(tf.shape(labels)[0])

        training_loss += float(loss) * batch_size
        training_samples += batch_size

    train_loss = training_loss / training_samples
    val_metrics = validate_model()

    print(
        f"Epoch {epoch}/{EPOCHS} | "
        f"Train loss: {train_loss:.4f} | "
        f"Val loss: {val_metrics['loss']:.4f} | "
        f"Civic recall: {val_metrics['civic_recall']:.4f} | "
        f"Non-civic recall: {val_metrics['noncivic_recall']:.4f}"
    )

    if val_metrics["loss"] < best_val_loss:
        best_val_loss = val_metrics["loss"]
        epochs_without_improvement = 0

        # Save locally first, then copy the best checkpoint to Drive
        binary_model_v2.save(LOCAL_BEST)
        shutil.copy2(LOCAL_BEST, DRIVE_BEST)

        print("  Best model checkpoint saved to Drive.")
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= PATIENCE:
        print("Early stopping: validation loss did not improve.")
        break

print("\nTraining finished.")
print("Best validation loss:", best_val_loss)
print("Best model:", DRIVE_BEST)
```

```python
import os
import shutil

MODEL_DIR = "/content/gdrive/MyDrive/CivicComplaintML/models"

# Source: best improved binary model
binary_source = os.path.join(
    MODEL_DIR, "urbanpulse_binary_v2_best.keras"
)

# Final binary model name
binary_final = os.path.join(
    MODEL_DIR, "urbanpulse_binary_relevance_v2.keras"
)

# Eight-class classifier
classifier_path = os.path.join(
    MODEL_DIR, "urbanpulse_8class_pothole_stage1.keras"
)

# Verify source models exist
assert os.path.exists(binary_source), "Improved binary model is missing!"
assert os.path.exists(classifier_path), "Eight-class model is missing!"

# Save improved binary model under its final name
shutil.copy2(binary_source, binary_final)

print("FINAL MODEL VERIFICATION")
print("-" * 45)

for path in [binary_final, classifier_path]:
    print("Model:", os.path.basename(path))
    print("Saved:", os.path.exists(path))
    print("Size:", round(os.path.getsize(path) / (1024 * 1024), 2), "MB")
    print("-" * 45)

print("Both UrbanPulse models are saved in Google Drive.")
```
